```python
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
# Task One

Bcrits = []
s_profs = []
omega_etas = []

fp = '/Users/paullab/codes/firm3d_fork_10132025/firm3d/examples/trapped_map/data01222026'
fns = sorted(os.listdir(fp))
num = len(fns)/3

fig,ax = plt.subplots()
j=0
for i in range(0,len(fns)):
    if fns[i][0] == 'B':
        Bcrit = fns[i][6:-4]
        Omega_eta_prof = np.load(fp+'/Omega_eta_arr'+Bcrit+'.npy',allow_pickle=True)
        s_prof = np.sqrt(np.load(fp+'/s_arr'+Bcrit+'.npy',allow_pickle=True))

        if Bcrit == '5.9' or Bcrit == '6.3': # lo oks like this one will have a crazy Poincare plot
            ax.plot(Omega_eta_prof,s_prof,label=r'$B_{crit} = $'+Bcrit)
            j+=1


# Resonance stuff #
# Rational numbers
p_max=10
q_max=4
res_range_min = -4
res_range_max = 4
res_arr = np.array([])
# q_arr = np.array([])
res_arr_set = 0

# including the zero resonance
res_arr = np.append(res_arr,0)

for p in range(1,p_max+1): # include the zero resonance
    for q in range(1,q_max+1):
            condition1 = p/q >= res_range_min and p/q <= res_range_max
            if ~np.isin(p/q, res_arr) and condition1:
                res_arr = np.append(res_arr,p/q)
                res_arr = np.append(res_arr,-p/q)

for res in res_arr:
    ax.plot([res,res],[0,1],'k--',lw=0.5)


plt.legend(bbox_to_anchor=(1.3, 1), loc='upper right')
ax.set_ylabel(r"$\rho$")
ax.set_xlabel(r"$\Omega_\zeta$")
ax.set_xlim([-0.5,0.5])

    
```
