```python
import numpy as np
import matplotlib.pyplot as plt
import time
import desc.io
from desc.ResonanceOpt.TRObj_func import TrappedResonanceObj
```

```python
# Import FIRM3D data
firm3d_times = np.load("times.npy")
firm3d_Ekin_fracs = np.load("Ekin_fracs.npy")
firm3d_neta_poincs = np.load("neta_poincs.npy")
firm3d_ns_poincs = np.load("ns_poincs.npy")
```

```python
# DESC loop
time_loop = range(0,10)
DESC_rhos = np.zeros(len(time_loop))
DESC_alphas = np.zeros(len(time_loop))
DESC_times = np.zeros(len(time_loop))
neta_poinc=3
Ekin_frac=0.1
for i in time_loop:
    # neta_poinc=i+2
    ns_poinc=(i+1)*10
    alphas = np.linspace(0,2*np.pi,neta_poinc)
    rhos = (np.linspace(0.1,0.9,ns_poinc))**0.5
    DESC_times[i],DESC_rhos[i],DESC_alphas[i] = TrappedResonanceObj(rhos=rhos,pitch_invs=np.array([5.95]),KE_frac=np.array([0.1]),alphas=alphas,N=0)
    print(DESC_times[i])
```
